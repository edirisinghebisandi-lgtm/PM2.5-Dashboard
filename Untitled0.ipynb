```python
import pandas as pd
df = pd.read_csv('/content/drive/MyDrive/Data science_individual/esp_pm2_5_source (1).csv', encoding='latin1')
df.head()
```

```python
df.info()
df.isnull().sum()
```

```python
df = df[['year', 'emissionsQuantity', 'name',
         'sector', 'subsector', 'centroid',
         'activity', 'capacity']]
```

```python
df = df.rename(columns={
    'emissionsQuantity': 'PM25_Emissions',
    'name': 'Region'
})
```

```python
df['longitude'] = df['centroid'].str.extract(r"'longitude': ([\-\d\.]+)").astype(float)
df['latitude'] = df['centroid'].str.extract(r"'latitude': ([\-\d\.]+)").astype(float)
```

```python
df = df.drop(columns=['centroid'], errors='ignore')
```

```python
df['year'] = df['year'].astype(int)
df['PM25_Emissions'] = pd.to_numeric(df['PM25_Emissions'], errors='coerce')
df['activity'] = pd.to_numeric(df['activity'], errors='coerce')
df['capacity'] = pd.to_numeric(df['capacity'], errors='coerce')
```

```python
#remove duplicates
df = df.drop_duplicates()
```

```python
#Remove missing values
df = df.dropna()
```

```python
#Remove zero values
df = df[~((df['PM25_Emissions'] == 0) & (df['activity'] == 0))]
```

```python
#Remove "Unknown name"
df = df[df['Region'] != "Unknown name"]
```

```python
#remove extreme outliers
df = df[df['PM25_Emissions'] > 0]
```

```python
#Reset index
df = df.reset_index(drop=True)
```

```python
df.to_csv("Cleaned_PM2.5_dataset.csv", index=False)
```
